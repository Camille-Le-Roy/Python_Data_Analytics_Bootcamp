## Parte 1: Que es un framework de gestion de proyectos de datos

Un framework de gestion de proyectos es una metodologia que define en que orden abordar las etapas de un proyecto, y que preguntas hay que responder en cada una. Un framework es un proceso que sigue el equipo.

Sin un framework, es comun caer en un patron como este:

```text
"Tenemos un dataset. Vamos a programar."
```

y recien varias semanas despues, a veces cuando ya hay un modelo entrenado, aparecen las preguntas que debieron responderse al principio: para que sirve esto, quien lo va a usar, como se mide si funciona. Responder esas preguntas tarde suele significar rehacer trabajo.

Un framework invierte el orden: obliga a responder las preguntas de negocio antes de escribir la primera linea de codigo de modelado.

## Parte 2: CRISP-DM

**CRISP-DM** (Cross-Industry Standard Process for Data Mining) es uno de los frameworks mas usados para estructurar proyectos de datos, sin importar la industria. Define seis fases:

```text
1. Business Understanding   -> Que problema de negocio estamos resolviendo?
2. Data Understanding       -> Que datos tenemos, y podemos confiar en ellos?
3. Data Preparation         -> Limpieza, transformacion, creacion de features
4. Modeling                 -> Que modelo o analisis vamos a aplicar?
5. Evaluation                -> El resultado realmente sirve para el negocio?
6. Deployment                -> Como se usa esto en la practica?
```

Dos cosas importantes sobre estas fases:

- **No son estrictamente lineales.** Es comun volver de Evaluation a Data Preparation, por ejemplo, si el modelo no funciona bien y hay que revisar los datos.
- **Business Understanding es la fase que mas se salta**, y es la que mas caro sale saltarse. Un modelo tecnicamente excelente que responde la pregunta equivocada no sirve de nada.

Veamos como se verian estas seis fases aplicadas a un caso concreto.

**Escenario:** Un supermercado quiere predecir que clientes van a dejar de comprar (esto se conoce como *churn*).

| Fase | Pregunta que responde | En este caso |
|---|---|---|
| Business Understanding | Que problema de negocio resolvemos? | Que significa "dejar de comprar" para el negocio? Con cuanta anticipacion se necesita saber? Que decision se toma con la prediccion (por ejemplo, enviar un descuento)? |
| Data Understanding | Que datos tenemos? | Historial de compras, frecuencia de visitas, categorias compradas. Cuanta historia hay? Faltan datos? |
| Data Preparation | Como preparamos los datos? | Definir la ventana de tiempo, limpiar valores faltantes, crear variables como "dias desde la ultima compra" |
| Modeling | Que modelo aplicamos? | Un modelo de clasificacion: el cliente se va o se queda |
| Evaluation | El resultado sirve? | El modelo predice mejor que simplemente asumir "nadie se va"? Las alertas llegan a tiempo para actuar? |
| Deployment | Como se usa en la practica? | Un dashboard semanal para el equipo de marketing? Una alerta automatica? |

Notemos que Modeling, la fase que suele entusiasmar mas a quien recien empieza en ciencia de datos, es apenas una de seis. Y depende totalmente de que las dos primeras esten bien resueltas.
