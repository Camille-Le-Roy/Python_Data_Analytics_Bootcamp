## Parte 1: Por que documentar el codigo

La documentacion tecnica permite que el codigo sea legible por otras personas, no solo por quien lo escribio. En Python, la forma estandar de documentar una funcion es con un **docstring**: un texto entre comillas triples justo debajo de la definicion de la funcion, que Python reconoce como parte del objeto funcion (se puede consultar con `.__doc__` o con `help()`).

Existen varios estilos de docstring. Nosotros vamos a usar el estilo **NumPy**, muy comun en el ecosistema de ciencia de datos (pandas, numpy y scikit-learn lo usan). Un docstring en este estilo generalmente responde tres preguntas:

- **Parameters**: que recibe la funcion, y de que tipo es cada parametro.
- **Returns**: que devuelve la funcion, y de que tipo.
- **Raises** (cuando aplica): que errores puede lanzar la funcion, y en que casos.

#### Ejemplo de funcion describido con el **Numpy style**

```python
def procesar_inventario(df, stock_minimo):
    """
    Filtra los productos que requieren reabastecimiento.

    Parameters
    ----------
    df : pandas.DataFrame
        Dataset con columnas ['producto', 'cantidad'].
    stock_minimo : int
        Umbral para alertar falta de inventario.

    Returns
    -------
    pandas.DataFrame
        Productos cuyo stock es menor al minimo.
    """
    return df[df['cantidad'] < stock_minimo]
```

El docstring no repite el codigo, explica la intencion: que hace la funcion, que espera recibir, y que devuelve.<br><br>

#### Ejemplo de funcion describido con el **Google style**

A diferencia del estilo NumPy, utiliza secciones como `Args`, `Returns` y `Raises`.

- Args:
    parametro (tipo): Descripción.

- Returns:
    tipo: Descripción.

- Raises:
    TipoDeError: Situación que genera el error.
  


## Parte 2: Documentando `calcular_metricas_ventas`

En la sesion 1 dejamos `src/procesamiento.py` con solo un comentario placeholder. Hoy le agregamos la funcion `calcular_metricas_ventas(dataframe)`, que recibe un dataset de ventas con columnas `producto`, `cantidad` y `precio_unitario`, y devuelve un diccionario con tres metricas: el total vendido, el ticket promedio, y la cantidad de transacciones.

Un detalle importante: la funcion debe comportarse bien incluso si el dataframe esta vacio (sin filas). En vez de lanzar un error, va a devolver las metricas en cero. Vamos a verificar este comportamiento con una prueba unitaria en la sesion 3.

Sobrescribimos `src/procesamiento.py` con la funcion completa y documentada.

````python
# Creamos el archivo procesamiento.py dentro de la carpeta src
%%writefile src/procesamiento.py

"""Modulo de procesamiento de metricas de ventas."""  # Describimos brevemente el objetivo del modulo

import pandas as pd  # Importamos Pandas para trabajar con DataFrames


def calcular_metricas_ventas(dataframe):
    """
    Calcula metricas clave a partir de un dataset de ventas.

    Parameters
    ----------
    dataframe : pandas.DataFrame
        Dataset de ventas con columnas ['producto', 'cantidad', 'precio_unitario'].

    Returns
    -------
    dict
        Diccionario con las llaves 'total_ventas', 'ticket_promedio' y
        'cantidad_transacciones'. Si el dataframe esta vacio, las tres
        metricas se devuelven en cero.
    """
    if dataframe.empty:  # Comprobamos si el DataFrame esta vacio (".empty" se puede usar sobre cualquier dataframe, series o inex de Pandas, it returns True or False)
        return {
            "total_ventas": 0.0,  # Si esta vacio, el total de ventas es cero
            "ticket_promedio": 0.0,  # Si esta vacio, el ticket promedio es cero
            "cantidad_transacciones": 0,  # Si esta vacio, no hay transacciones
        }

    ventas = dataframe.copy()  # Creamos una copia para no modificar el DataFrame original
    ventas["venta_total"] = ventas["cantidad"] * ventas["precio_unitario"]  # Calculamos el valor total de cada venta

    total_ventas = round(ventas["venta_total"].sum(), 2)  # Sumamos las ventas y redondeamos a dos decimales
    cantidad_transacciones = len(ventas)  # Contamos el numero de transacciones
    ticket_promedio = round(total_ventas / cantidad_transacciones, 2)  # Calculamos el promedio de venta por transaccion

    return {
        "total_ventas": total_ventas,  # Devolvemos el total de ventas
        "ticket_promedio": ticket_promedio,  # Devolvemos el ticket promedio
        "cantidad_transacciones": cantidad_transacciones,  # Devolvemos el numero de transacciones
    }
````

Probemos la funcion con un dataset pequeno. La importamos desde `src.procesamiento`, tal como lo haria cualquier otro archivo del proyecto.

````python
from src.procesamiento import calcular_metricas_ventas # Importamos la función de procesamiento desde nuestro módulo

df_demo = pd.DataFrame({ # Creamos un DataFrame de ejemplo con datos de ventas
    "producto": ["A", "B", "C"],
    "cantidad": [2, 1, 5],
    "precio_unitario": [10.0, 25.0, 4.0],
})

calcular_metricas_ventas(df_demo) # Calculamos las métricas de ventas utilizando el DataFrame de ejemplo
````

## Parte 3: Type hints

Además del docstring, Python permite indicar los tipos de datos esperados directamente en la función.

Los **type hints** ayudan a entender qué tipo de información recibe y devuelve una función. También permiten que editores y herramientas detecten posibles errores.



### Ejemplo sin type hints

```python
def calcular_total(precio, cantidad):
    return precio * cantidad
```
La función funciona, pero no indica claramente qué tipo de dato espera cada parámetro ni qué devuelve

### Ejemplo con type hints

```python
def calcular_total(precio: float, cantidad: int) -> float:
    return precio * cantidad
```
Los type hints sirven principalmente como **información para el programador y las herramientas de desarrollo**. Python no bloquea automáticamente un valor de otro tipo.

Ejemplo de una funcion con varias arguments:  
```python
def procesar_venta(
    producto: str,
    cantidad: int,
    precio: float,
    disponible: bool
) -> float:
    return cantidad * precio
```


## Documenta esta funcion

La siguiente funcion calcula un descuento sobre un monto, pero no tiene docstring. Escribe tu propio docstring en estilo NumPy antes de continuar (Parameters, Returns, y Raises si corresponde).

````python
def calcular_descuento(monto, porcentaje):
    """
    Que hace la funcion? Que recibe?
    Que devuelve? Y que pasa si el porcentaje esta fuera de rango?
    """
    if not 0 <= porcentaje <= 100:
        raise ValueError("El porcentaje debe estar entre 0 y 100")
    return round(monto * (1 - porcentaje / 100), 2)


help(calcular_descuento)
````

````python
def calcular_descuento(monto, porcentaje):

    """
    Calcula el monto final después de aplicar un descuento.

    Parameters
    ----------
    monto : float
        Monto original al que se aplicará el descuento.
    porcentaje : float
        Porcentaje de descuento, debe estar entre 0 y 100.

    Returns
    -------
    float
        Monto final después de aplicar el descuento, redondeado a dos decimales.

    Raises
    ------
    ValueError
        Si el porcentaje está fuera del rango de 0 a 100.
    """

    if not 0 <= porcentaje <= 100:
        raise ValueError("El porcentaje debe estar entre 0 y 100")

    return round(monto * (1 - porcentaje / 100), 2)


help(calcular_descuento)
````

## Errores comunes

- **Docstring desactualizado**: cambiar el comportamiento de la funcion y olvidar actualizar el docstring. Es peor que no tener docstring, porque genera una falsa confianza.
- **Docstring demasiado vago**: escribir algo como "Calcula ventas" sin especificar que recibe ni que devuelve, no ayuda a nadie.
- **Repetir el codigo en palabras**: un buen docstring explica la intencion, no traduce linea por linea lo que ya se puede leer en el codigo.
