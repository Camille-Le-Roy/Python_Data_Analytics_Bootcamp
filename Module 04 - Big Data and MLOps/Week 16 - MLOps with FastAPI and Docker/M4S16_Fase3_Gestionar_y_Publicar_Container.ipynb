## Parte 1: Gestion de contenedores

Hasta ahora solo usamos `docker build` y `docker run`. Pero en el dia a dia, tambien hay que poder ver que esta corriendo, revisar que esta pasando adentro, y apagar o borrar lo que ya no se necesita.

Antes de empezar, corre un contenedor en segundo plano para tener algo que gestionar:

```
docker run -d -p 8000:8000 --name mi-api ml-api-app
```

El flag `-d` lo corre en segundo plano (detached) y `--name` le pone un nombre facil de recordar, en vez de tener que usar el ID largo que genera Docker.

Comandos para probar:

- `docker ps` -> lista los contenedores que estan corriendo ahora mismo. Mostrar el ID, el nombre, y el estado.
- `docker ps -a` -> lista **todos** los contenedores, incluidos los que ya se detuvieron. Util para ver el historial.
- `docker logs mi-api` -> muestra la salida de la aplicacion (lo que imprimiria uvicorn en la terminal si lo hubieras corrido directo). Sirve muchisimo para diagnosticar errores.
- `docker stop mi-api` -> detiene el contenedor, pero no lo borra. Se puede volver a arrancar despues.
- `docker start mi-api` -> vuelve a arrancar un contenedor detenido.
- `docker rm mi-api` -> borra el contenedor (tiene que estar detenido primero).
- `docker images` -> lista las imagenes que tenes construidas o descargadas en tu computadora.
- `docker rmi ml-api-app` -> borra una imagen (no debe haber contenedores usandola).


## Parte 2: Comparar tamanio de imagen (slim vs. full)

En la Sesion 2 mencionamos que usar una imagen base `-slim` es una buena practica. Verificamos con numeros reales.

Pasos:

1. Duplicar el Dockerfile de la API y en la copia cambiar la primera linea de `FROM python:3.9-slim` a `FROM python:3.9` (la version completa, sin `slim`). Se puede guardar como `Dockerfile.full`.
2. Construir las dos imagenes con nombres distintos:

```
docker build -t ml-api-app:slim -f Dockerfile .
docker build -t ml-api-app:full -f Dockerfile.full .
```

3. Comparar con:

```
docker images
```

4. Mirar la columna `SIZE` para las dos filas (`ml-api-app:slim` y `ml-api-app:full`). La diferencia suele ser de varios cientos de MB.



## Parte 3: Publicar en Docker Hub

Hasta ahora la imagen solo existe en tu computadora. Docker Hub es un repositorio en la nube (equivalente a lo que es GitHub para codigo) donde se pueden publicar imagenes para que cualquiera las descargue y corra, sin tener que compartir el codigo fuente directamente.

Requisito previo: tener una cuenta gratuita en https://hub.docker.com

Pasos:

1. Iniciar sesion desde la terminal:

```
docker login
```

Va a pedir usuario y contrasenia de Docker Hub.

2. Docker Hub exige que las imagenes se nombren como `usuario/nombre-imagen`. Como ya construimos la imagen con otro nombre, hay que "etiquetarla" de nuevo (no la reconstruye, solo le agrega un alias):

```
docker tag ml-api-app tu-usuario/ml-api-app:v1
```

(Reemplazar `tu-usuario` por el nombre de usuario real de Docker Hub.)

3. Publicar la imagen:

```
docker push tu-usuario/ml-api-app:v1
```

4. Mostrar el resultado entrando a `https://hub.docker.com/r/tu-usuario/ml-api-app` en el navegador: ahi queda visible la imagen, lista para que cualquiera la descargue con `docker pull tu-usuario/ml-api-app:v1` y la corra en su propia computadora, sin haber visto nunca el codigo fuente.
